# Data Cleaning & Feature Engineering

```python
df_clean = df.copy()
```

```python
df_clean.drop(
    columns=[
        "job_id",
        "salary_range",
        "department"
    ],
    inplace=True
)
```

```python
df_clean.columns
```

```python
df_clean.duplicated().sum()
```

```python
df_clean.drop_duplicates(inplace=True)
```

```python
# Fill missing values only for the fields used by the final model

text_columns = [
    "title",
    "description"
]

for col in text_columns:
    df_clean[col] = df_clean[col].fillna("")

binary_columns = [
    "telecommuting",
    "has_company_logo",
    "has_questions"
]

for col in binary_columns:
    df_clean[col] = df_clean[col].fillna(0)
```

```python
df_clean.isnull().sum()
```

```python
df_clean.duplicated().sum()
```

```python
df_clean.drop_duplicates(inplace=True)
```

```python
# Create the exact text input available to the deployed application

df_clean["text"] = (
    df_clean["title"] + " " +
    df_clean["description"]
)
```

```python
df_clean["text"].head()
```

```python
df_clean = df_clean[
    [
        "text",
        "telecommuting",
        "has_company_logo",
        "has_questions",
        "fraudulent"
    ]
]
```

```python
df_clean.head()
```

```python
import nltk

nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
import re

from nltk.corpus import stopwords

from nltk.stem import WordNetLemmatizer

# Create the objects
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()
```

```python
def clean_text(text):

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+", "", text)

    # Remove HTML
    text = re.sub(r"<.*?>", "", text)

    # Remove punctuation and numbers
    text = re.sub(r"[^a-zA-Z ]", " ", text)

    # Tokenize
    words = text.split()

    # Remove stopwords and lemmatize
    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    return " ".join(words)
```

```python
df_clean["text"] = df_clean["text"].apply(clean_text)
```

```python
df_clean["text"].head()
```

```python
from sklearn.model_selection import train_test_split
from scipy.sparse import hstack

# Text input
X_text_raw = df_clean["text"]

# Binary inputs
X_binary_raw = df_clean[
    [
        "telecommuting",
        "has_company_logo",
        "has_questions"
    ]
]

# Target
y = df_clean["fraudulent"]
```

```python
X_train_text, X_test_text, X_train_binary, X_test_binary, y_train, y_test = train_test_split(
    X_text_raw,
    X_binary_raw,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

# Feature Engineering (Vectorization)

This is where we convert text into numbers, because ML algorithms cannot understand raw text.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    min_df=3
)
```

```python
X_train_text_tfidf = tfidf.fit_transform(X_train_text)

X_test_text_tfidf = tfidf.transform(X_test_text)
```

```python
X_train = hstack([
    X_train_text_tfidf,
    X_train_binary.values
])

X_test = hstack([
    X_test_text_tfidf,
    X_test_binary.values
])
```

# Model Training

```python
from sklearn.naive_bayes import MultinomialNB

from sklearn.linear_model import LogisticRegression

from sklearn.svm import LinearSVC
```

```python
models = {

    "Naive Bayes": MultinomialNB(),

    "Logistic Regression": LogisticRegression(
        max_iter=1000
    ),

    "Linear SVM": LinearSVC(),

    "Balanced Linear SVM": LinearSVC(
        class_weight="balanced"
    )

}
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
results = []
```

```python
for name, model in models.items():

    print("=" * 60)
    print(name)

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, zero_division=0)
    recall = recall_score(y_test, y_pred, zero_division=0)
    f1 = f1_score(y_test, y_pred, zero_division=0)

    results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Fraud Precision": precision,
        "Fraud Recall": recall,
        "Fraud F1": f1
    })

    print(
        classification_report(
            y_test,
            y_pred,
            target_names=["Genuine", "Fraudulent"],
            zero_division=0
        )
    )
```

```python
results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Fraud F1",
    ascending=False
)

results_df
```

```python
best_name = results_df.iloc[0]["Model"]

best_model = models[best_name]

y_best_pred = best_model.predict(X_test)

cm = confusion_matrix(y_test, y_best_pred)

print("Best model:", best_name)
print("\nConfusion Matrix:")
print(cm)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_best_pred,
        target_names=["Genuine", "Fraudulent"],
        zero_division=0
    )
)
```

```python
# lOAD Models
model = joblib.load("../models/fake_job_detector.pkl")

vectorizer = joblib.load("../models/tfidf_vectorizer.pkl")
```

```python
from scipy.sparse import hstack

def predict_job(
    title,
    description,
    telecommuting=0,
    has_company_logo=0,
    has_questions=0
):

    # Combine exactly the same fields used during training
    text = f"{title} {description}"

    # Apply the same preprocessing used during training
    text = clean_text(text)

    # Transform using the TF-IDF fitted only on training data
    text_vector = vectorizer.transform([text])

    # Add the three binary inputs
    binary = [[
        telecommuting,
        has_company_logo,
        has_questions
    ]]

    # Combine text + binary features
    vector = hstack([
        text_vector,
        binary
    ])

    prediction = model.predict(vector)[0]

    return prediction
```

```python
sample_title = "Software Engineer"

sample_description = """
Google is hiring a Software Engineer.

Bachelor degree required.

Medical insurance.

401K.

5 years experience.
"""

predict_job(
    sample_title,
    sample_description,
    telecommuting=0,
    has_company_logo=1,
    has_questions=1
)
```

```python
sample2_title = "Work From Home Opportunity"

sample2_description = """
Earn $5000 weekly from home.

No experience required.

Click this WhatsApp number.

Immediate joining.

Limited seats.
"""

predict_job(
    sample2_title,
    sample2_description,
    telecommuting=1,
    has_company_logo=0,
    has_questions=0
)
```
